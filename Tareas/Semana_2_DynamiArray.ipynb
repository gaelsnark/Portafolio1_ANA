```python
import ctypes                                     # provides low-level arrays

class DynamicArray:
    """A dynamic array class akin to a simplified Python list."""

    def __init__(self):
        """Create an empty array."""
        self._n = 0                               # count actual elements
        self._capacity = 1                        # default array capacity
        self._A = self._make_array(self._capacity) # low-level array

    def __len__(self):
        """Return number of elements stored in the array."""
        return self._n

    def __getitem__(self, k):
        """Return element at index k."""
        if not 0 <= k < self._n:
            raise IndexError('invalid index')
        return self._A[k]                         # retrieve from array

    def append(self, obj):
        """Add object to end of the array."""
        if self._n == self._capacity:              # not enough room
            self._resize(2 * self._capacity)      # so double capacity
        self._A[self._n] = obj
        self._n += 1

    def _resize(self, c):                         # nonpublic utility
        """Resize internal array to capacity c."""
        B = self._make_array(c)                   # new (bigger) array
        for k in range(self._n):                  # for each existing value
            B[k] = self._A[k]
        self._A = B                               # use the bigger array
        self._capacity = c

    def _make_array(self, c):                     # nonpublic utility
        """Return new array with capacity c."""
        return (c * ctypes.py_object)( )          # see ctypes documentation
```

```python
import time 
# Time es: Unalibrería estándar que permite manejar funciones relacionadas con la hora, medir tiempos de ejecución y pausar programas

data = [100, 1000, 10000, 100000, 1000000]

for N in data:
    
    # Caso 1 pop al inicio: k=0
    data = [None] * N
    inicio = time.perf_counter() # perf_counter() se usa para medir el tiempo exacto que tarda en ejecutarse un fragmento de código (lo que se conoce como hacer un benchmark o prueba de rendimiento).
    for i in range(N):
        data.pop(0)
    fin = time.perf_counter()
    promedio = (fin - inicio) / N * 1000000
    print(f"k=0, N={N}: {promedio:.3f} microsegundos")
    
    # caso 2: pop en el medio k=//2
    data = [None] * N
    inicio = time.perf_counter()
    for i in range(N):
        mitad = len(data) // 2
        data.pop(mitad)
    fin = time.perf_counter()
    promedio = (fin - inicio) / N * 1000000
    print(f"k=n//2, N={N}: {promedio:.3f} microsegundos")
    
    # Caso 3 pop al final (k=n-1) 
    data = [None] * N
    inicio = time.perf_counter()
    for i in range(N):
        data.pop()
    fin = time.perf_counter()
    promedio = (fin - inicio) / N * 1000000
    print(f"k=n-1, N={N}: {promedio:.3f} microsegundos")
    
    print()  # línea en blanco entre tamaños
```

**C-5.14** The shuffle method, supported by the random module, takes a Python list and rearranges it so that every possible ordering is equally likely. Implement your own version of such a function. You may rely on the randrange(n) function of the random module, which returns a random number between 0 and n−1 inclusive.

```python
import random

def Shuffle_list(data1):
    for i in range(len(data1) - 1, 0, -1):
        # randrange(i + 1) nos da un número de 0 a i
        j = random.randrange(i + 1)
        data1[i],data1[j] = data1[j],data1[i]

data = [random.randint(1, 90) for _ in range(58)]

print(f"Lista original: {data}")
Shuffle_list(data)
print(f"Lista con shuffle: {data}")
```

C-5.15 Consider an implementation of a dynamic array, but instead of copying the elements into an array of double the size (that is, from N to 2N) when its capacity is reached, we copy the elements into an array with ⌈N/4⌉ additional cells, going from capacity N to capacity N+⌈N/4⌉. Prove that performing a sequence of n append operations still runs in O(n) time in this case.

### Lo que está pasando
Una caja es un array, donde guardas cosas una por una. Eventualmente se llena y al llenarse debes comprar una nueva caja y pasar el contenido a esta nueva caja. Esto conlleva tiempo porque como va de una en una, se tarda. 
La idea es si en vez de duplicar (de N a 2N), solo la agrandas de poco a poco ($1/4$ cada vez más grande) 

O(N) quiere decir un elemento de entrada equivale a una unidad de trabajo.

Imaginemos que tenemos una caja, y cuando agregamos un objeto a la caja pagamos 6$
* Pero aquí de esos 6 se paga solo 1 realmente para agregar el objeto.
* Los otros 5 se guardan
  
Entonces así vamos ahorrando cada vez que queramos agregar un objeto a la caja. 
Cuando la caja se llena y se quiera agrandar y mover los objetos usamos ese dinero ahorrado

¿Por qué 6 monedas?
Porque con el sistema duplicar al agrandar la caja la mitad de los espacios son nuevos, así cada moneda ahorra 1 extra esto hace
que el total sean 3 monedas por objeto.
Pero usando el $ 1/4$ solo 1 de cada 5 espacios es nuevo. Así cada moneda tiene que ahorrar más, 5 monedas en vez de 1 para que alcance el dinero
cuando necesitemos agrandar de nuevo.

Como cada inserción cuesta una cantidad fija de monedas (6), y nunca te quedas sin dinero ahorrado para pagar los agrandamientos, entonces:

$ n inserciones×6 monedas=6n=O(n) $

Es decir, hacer n inserciones cuesta, en total, un tiempo proporcional a n — no se vuelve lento aunque agrandes la caja poquito a poquito.
Esto es amortizar.


Operación:

=> $N=N'+⌈N'/4⌉⟹N'≥4/5​N$


=>$ N−N'≥N−4/5​N=1/5​N. $

=> $ 1/5​N⋅5=N $

-> $ 6n=O(n). $
