### Ejercicio 4
Invertir una lista sin usar ⁠ reverse() ⁠ ni slicing inverso.

```python
nug1=int(input("Ingresa cuántos números quieres: "))
lis=[]
invertida=[]

for i in range(nug1):
    nums=int((input("Ingresa los números de la lista: ")))
    lis.append(nums)

for inv in range(len(lis)-1,-1,-1):
    invertida.append(lis[inv])
    print(invertida)
```

### Ejercicio 7

### Ejercicio 8
Implementar búsqueda binaria en lista ordenada.

```python

lis = []

cant = int(input("Rango de la lista: "))
for i in range(cant):
    valor = int(input(f"Valor {i + 1} de la lista: "))
    lis.append(valor)

objetivo = int(input(f"Ingrese el número para encontrar: "))
izquierda = 0
derecha=len(lis)-1

# Ordenar la lista antes de realizar la búsqueda binaria
lis.sort()
print("Lista ordenada:", lis)

while izquierda <= derecha:
    medio = (izquierda + derecha) // 2  # Encuentra el índice del medio
    if lis[medio] == objetivo:
        print(f"Valor medio: {medio}")
        break
    elif lis[medio] < objetivo:
        izquierda = medio + 1  # Busca en la mitad derecha
        print(f"Por la mitad derecha, nueva izquierda es: {izquierda}")
    else:
        derecha = medio - 1  # Busca en la mitad izquierda
        print(f"Por la mitad izquierda, nueva derecha es: {derecha}")

if izquierda > derecha:
    print(f"Objetivo no encontrado en la lista")
```

### Ejercicio 9
Eliminar duplicados manteniendo el orden.

```python
cant = int(input("Rango de la lista: "))
lista = [] * cant
for i in range(cant):
    valor = int(input(f"Valor {i + 1} de la lista: "))
    lista.append(valor)

resultado = []
for x in lista:
    if x not in resultado:
        resultado.append(x)
print(f"Lista sin repetidos: {resultado}")
```

### Ejercicio 10

### Ejercicio 12

```python
lista1 = [1, 2, 3]

clonar = lista1.copy()
slicing = lista1[:]

# Mostrar independencia
clonar[0]=5858
slicing.append(4648)

print("Lista 1: ", lista1)
print("Lista con copy (cambia su índice 0): ", clonar)
print("Lista con slicing (se le agrega): ", slicing)
```

### Ejercicio 13
Implementar una función que intercambie dos elementos de una lista por referencia.

```python
def Intercambiar_ref(lista,indice1,indice2):
    lista[indice1],lista[indice2] = lista[indice2], lista[indice1]

OthList = [5,6,8,5,8]
print(f"La lista antes de intercambiar los números: {OthList}")

Intercambiar_ref(OthList,1,9)

print(f"Lista después de intercambiar: {OthList}")
```
